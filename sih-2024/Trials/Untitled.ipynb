```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
import os
DATA_DIR = os.environ.get("DATA_DIR", "../../data")
df = pd.read_csv(os.path.join(DATA_DIR, "Delhi_2022_PowerVSTemp.csv"),index_col="Date",parse_dates = True)
df = df.dropna()
print("Shape", df.shape)
df.head()
```

```python
df["Peak Demand (in MW)"].plot(figsize = (12,5))
plt.legend()
df["Maximum Temprature (in °C)"].plot(figsize = (12,6))
plt.legend()
df["Minimum Temprature (in °C)"].plot(figsize = (12,6))
plt.legend()
```
