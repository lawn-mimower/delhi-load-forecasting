```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from IPython.display import display
from IPython.display import HTML
```

```python
import os
DATA_DIR = os.environ.get("DATA_DIR", "../data")
df = pd.read_csv(os.path.join(DATA_DIR, "final_modified.csv"),parse_dates = ["Date and Time"])
print(df)
```

```python
plt.figure(figsize=(50,6))
plt.plot(df["Date and Time"],df["temp"])
plt.savefig("TempVsDate.png")
plt.show()

display(HTML('''
    <div style="width:100%; overflow-x: scroll; white-space: nowrap;">
        <img src="TempVsDate.png" style="display: inline-block;">
    </div>
'''))
```
